```python
# ============================================================
# PROJECT 10: BANKING TRANSACTION PATTERN ANALYSIS
# ============================================================

import pandas as pd
import numpy as np
import re
from google.colab import files
```

```python
# ------------------------------------------------------------
# 1. LOAD AND INSPECT DATASET
# ------------------------------------------------------------

print("Upload 10_bank_transactions.csv")
uploaded = files.upload()

file_name = next(iter(uploaded))
df = pd.read_csv(file_name)

print("\nFIRST 5 RECORDS:")
display(df.head())

print("\nDATASET INFORMATION:")
print(df.info())

print("\nDATASET SHAPE:", df.shape)

print("\nORIGINAL COLUMN NAMES:")
print(df.columns.tolist())
# ------------------------------------------------------------
# STANDARDIZE COLUMN NAMES
# ------------------------------------------------------------

def normalize_name(name):
    return re.sub(r"[^a-z0-9]", "", str(name).lower())

aliases = {
    "Transaction_ID": [
        "transaction_id", "transactionid", "txn_id", "id"
    ],
    "Transaction_Amount": [
        "transaction_amount", "amount", "transaction_value",
        "txn_amount"
    ],
    "Transaction_Type": [
        "transaction_type", "type", "txn_type"
    ],
    "Channel": [
        "channel", "transaction_channel", "payment_channel"
    ],
    "City": [
        "city", "transaction_city", "location"
    ],
    "Flagged": [
        "flagged", "is_flagged", "flag", "alert_flag"
    ],
    "Account_Type": [
        "account_type", "accounttype", "account"
    ]
}

available_columns = {
    normalize_name(col): col for col in df.columns
}

rename_columns = {}

for standard_name, possible_names in aliases.items():
    candidates = [standard_name] + possible_names

    for name in candidates:
        key = normalize_name(name)

        if key in available_columns:
            rename_columns[available_columns[key]] = standard_name
            break

df.rename(columns=rename_columns, inplace=True)

required_columns = list(aliases.keys())
missing_columns = [
    col for col in required_columns if col not in df.columns
]

if missing_columns:
    raise ValueError(
        "Required columns not found: "
        + str(missing_columns)
        + "\nAvailable columns: "
        + str(df.columns.tolist())
    )

print("\nSTANDARDIZED COLUMN NAMES:")
print(df.columns.tolist())
```

```python
# ------------------------------------------------------------
# 2. FIND AND HANDLE MISSING TRANSACTION AMOUNTS
# ------------------------------------------------------------

df["Transaction_Amount"] = pd.to_numeric(
    df["Transaction_Amount"]
    .astype("string")
    .str.replace(r"[₹,$,\s]", "", regex=True),
    errors="coerce"
)

missing_amounts = df["Transaction_Amount"].isnull().sum()

print("\nMISSING TRANSACTION AMOUNTS:", missing_amounts)

if df["Transaction_Amount"].notna().sum() == 0:
    raise ValueError("All transaction amounts are missing.")

median_amount = df["Transaction_Amount"].median()

df["Transaction_Amount"] = df[
    "Transaction_Amount"
].fillna(median_amount)

print("Missing amounts handled using median:", median_amount)

```

```python
# ------------------------------------------------------------
# 3. CHECK DUPLICATE TRANSACTION IDs
# ------------------------------------------------------------

duplicate_mask = (
    df["Transaction_ID"].notna()
    & df["Transaction_ID"].duplicated(keep=False)
)

duplicate_ids = df.loc[
    duplicate_mask, "Transaction_ID"
].nunique()

duplicate_rows = duplicate_mask.sum()

print("\nDUPLICATE TRANSACTION IDs:", duplicate_ids)
print("ROWS WITH DUPLICATE IDs:", duplicate_rows)

# Keep the first record for each non-missing transaction ID
remove_duplicate_mask = (
    df["Transaction_ID"].notna()
    & df["Transaction_ID"].duplicated(keep="first")
)

df = df.loc[~remove_duplicate_mask].copy()

print("Records after duplicate handling:", len(df))


# ------------------------------------------------------------
# PREPARE DATA FOR ANALYSIS
# ------------------------------------------------------------

def normalize_transaction_type(value):
    if pd.isna(value):
        return "Unknown"

    value = str(value).strip().lower()

    mapping = {
        "credit": "Credit",
        "cr": "Credit",
        "deposit": "Credit",
        "debit": "Debit",
        "dr": "Debit",
        "withdrawal": "Debit"
    }

    return mapping.get(value, value.title())


def normalize_flag(value):
    if pd.isna(value):
        return pd.NA

    value = str(value).strip().lower()

    if value in ["1", "true", "yes", "y", "flagged"]:
        return True

    if value in [
        "0", "false", "no", "n", "not flagged",
        "unflagged", "normal"
    ]:
        return False

    return pd.NA


df["Transaction_Type"] = df[
    "Transaction_Type"
].apply(normalize_transaction_type)

df["Flagged"] = pd.Series(
    [normalize_flag(value) for value in df["Flagged"]],
    index=df.index,
    dtype="boolean"
)

for col in ["Channel", "City", "Account_Type"]:
    df[col] = (
        df[col].fillna("Unknown").astype("string").str.strip()
    )

# Amount magnitude is used for transaction-value comparisons.
# This also handles datasets where debits are stored as negative.
df["Amount_Abs"] = df["Transaction_Amount"].abs()
```

```python
# ------------------------------------------------------------
# 4. CALCULATE TOTAL CREDITS AND TOTAL DEBITS
# ------------------------------------------------------------

total_credits = df.loc[
    df["Transaction_Type"] == "Credit",
    "Amount_Abs"
].sum()

total_debits = df.loc[
    df["Transaction_Type"] == "Debit",
    "Amount_Abs"
].sum()

print("\nTOTAL CREDIT VALUE:", total_credits)
print("TOTAL DEBIT VALUE:", total_debits)

```

```python
# ------------------------------------------------------------
# 5. MEAN, MEDIAN AND STANDARD DEVIATION
# ------------------------------------------------------------

mean_amount = df["Amount_Abs"].mean()
median_transaction = df["Amount_Abs"].median()
std_deviation = df["Amount_Abs"].std()

print("\nTRANSACTION STATISTICS")
print("Mean:", mean_amount)
print("Median:", median_transaction)
print("Standard Deviation:", std_deviation)
```

```python
# ------------------------------------------------------------
# 6. COMPARE TRANSACTION VOLUME BY CHANNEL
# ------------------------------------------------------------

channel_summary = df.groupby("Channel").agg(
    Transaction_Count=("Transaction_ID", "size"),
    Total_Value=("Amount_Abs", "sum"),
    Average_Value=("Amount_Abs", "mean")
).reset_index()

channel_summary = channel_summary.sort_values(
    "Transaction_Count", ascending=False
)

print("\nTRANSACTION CHANNEL SUMMARY:")
display(channel_summary)
```

```python
# ------------------------------------------------------------
# 7. TOTAL TRANSACTION VALUE BY CITY
# ------------------------------------------------------------

city_summary = df.groupby("City").agg(
    Transaction_Count=("Transaction_ID", "size"),
    Total_Transaction_Value=("Amount_Abs", "sum")
).reset_index()

city_summary = city_summary.sort_values(
    "Total_Transaction_Value", ascending=False
)

print("\nCITY TRANSACTION SUMMARY:")
display(city_summary)
```

```python
# ------------------------------------------------------------
# 8. FLAGGED TRANSACTION COUNT AND PERCENTAGE
# ------------------------------------------------------------

valid_flag_count = int(df["Flagged"].notna().sum())

flagged_count = int(df["Flagged"].eq(True).sum())

flagged_percentage = (
    (flagged_count / valid_flag_count) * 100
    if valid_flag_count > 0 else 0
)

flagged_summary = pd.DataFrame({
    "Metric": [
        "Total Transactions",
        "Valid Flag Labels",
        "Flagged Transactions",
        "Flagged Percentage"
    ],
    "Value": [
        len(df),
        valid_flag_count,
        flagged_count,
        round(flagged_percentage, 2)
    ]
})

print("\nFLAGGED TRANSACTION SUMMARY:")
display(flagged_summary)

```

```python
# ------------------------------------------------------------
# 9. FLAGGED PERCENTAGE BY CHANNEL
# ------------------------------------------------------------

channel_flags = df.groupby("Channel").agg(
    Total_Transactions=("Transaction_ID", "size"),
    Flagged_Count=("Flagged", lambda x: x.eq(True).sum()),
    Valid_Flag_Labels=("Flagged", "count")
).reset_index()

channel_flags["Flagged_Percentage"] = np.where(
    channel_flags["Valid_Flag_Labels"] > 0,
    channel_flags["Flagged_Count"]
    / channel_flags["Valid_Flag_Labels"] * 100,
    np.nan
)

print("\nFLAGGED PERCENTAGE BY CHANNEL:")
display(channel_flags)
```

```python
# ------------------------------------------------------------
# 10. FIND THE 10 LARGEST TRANSACTIONS
# ------------------------------------------------------------

largest_transactions = df.nlargest(
    10, "Amount_Abs"
)

print("\n10 LARGEST TRANSACTIONS:")

display(
    largest_transactions[
        [
            "Transaction_ID",
            "Transaction_Type",
            "Transaction_Amount",
            "Channel",
            "City"
        ]
    ]
)
```

```python
# ------------------------------------------------------------
# 11. CREDIT AND DEBIT TOTALS BY ACCOUNT TYPE
# ------------------------------------------------------------

account_summary = df.groupby(
    ["Account_Type", "Transaction_Type"]
)["Amount_Abs"].sum().unstack(fill_value=0)

print("\nCREDIT AND DEBIT TOTALS BY ACCOUNT TYPE:")
display(account_summary)
```

```python
# ------------------------------------------------------------
# 12. IDENTIFY OUTLIERS USING IQR
# ------------------------------------------------------------

q1 = np.percentile(df["Amount_Abs"], 25)
q3 = np.percentile(df["Amount_Abs"], 75)

iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = df[
    (df["Amount_Abs"] < lower_bound)
    | (df["Amount_Abs"] > upper_bound)
]

print("\nIQR OUTLIER ANALYSIS")
print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Outliers:", len(outliers))

display(outliers.head(10))
```

```python
# ------------------------------------------------------------
# 13. CREATE TRANSACTION AMOUNT BANDS
# ------------------------------------------------------------

p25, p50, p75 = np.percentile(
    df["Amount_Abs"], [25, 50, 75]
)

df["Amount_Band"] = np.select(
    [
        df["Amount_Abs"] <= p25,
        df["Amount_Abs"] <= p50,
        df["Amount_Abs"] <= p75
    ],
    [
        "Low",
        "Medium",
        "High"
    ],
    default="Very High"
)

amount_band_summary = df.groupby(
    "Amount_Band", observed=False
).agg(
    Transaction_Count=("Transaction_ID", "size"),
    Total_Value=("Amount_Abs", "sum")
).reindex(
    ["Low", "Medium", "High", "Very High"]
).reset_index()

print("\nTRANSACTION AMOUNT BANDS:")
display(amount_band_summary)
```

```python
# ------------------------------------------------------------
# 15.display(amount_band_summary)
# FIVE CAUTIOUS OPERATIONAL OBSERVATIONS
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("OPERATIONAL OBSERVATIONS AND RECOMMENDATIONS")
print("=" * 60)

# Observation 1: Channel
if not channel_summary.empty:
    top_channel = channel_summary.iloc[0]

    print(
        "\n1. CHANNEL PERFORMANCE:"
        f"\n{top_channel['Channel']} has the highest "
        f"transaction volume with "
        f"{int(top_channel['Transaction_Count'])} records."
        "\nRecommendation: Review channel capacity, "
        "availability and transaction processing performance."
    )
# Observation 3: Flagged records
print(
    "\n3. FLAGGED RECORDS:"
    f"\n{flagged_count} transactions are flagged out of "
    f"{valid_flag_count} records with valid labels "
    f"({flagged_percentage:.2f}%)."
    "\nRecommendation: Review the rules that generate "
    "these labels and validate records individually. "
    "A flagged label is not proof of fraud."
)

# Observation 4: Account type
if not account_summary.empty:
    debit_by_account = (
        account_summary["Debit"]
        if "Debit" in account_summary.columns
        else pd.Series(dtype=float)
    )

    if not debit_by_account.empty:
        top_debit_account = debit_by_account.idxmax()

        print(
            "\n4. ACCOUNT TYPE:"
            f"\n{top_debit_account} has the highest "
            f"recorded debit value "
            f"({debit_by_account.max():.2f})."
            "\nRecommendation: Compare credit and debit "
            "patterns across account types and review "
            "whether the differences are expected."
        )
    else:
        print(
            "\n4. ACCOUNT TYPE: Debit records are unavailable "
            "for comparison."
            "\nRecommendation: Verify the transaction-type labels."
        )

# Observation 5: Outliers
print(
    "\n5. TRANSACTION OUTLIERS:"
    f"\nThe IQR method identified {len(outliers)} "
    "outlier records."
    "\nRecommendation: Review these records with "
    "transaction context and data-quality checks. "
    "An outlier is not automatically suspicious."
)

# ------------------------------------------------------------
# SAVE OUTPUT TABLES
# ------------------------------------------------------------

channel_summary.to_csv("channel_summary.csv", index=False)
city_summary.to_csv("city_summary.csv", index=False)
flagged_summary.to_csv("flagged_summary.csv", index=False)
channel_flags.to_csv("channel_flag_summary.csv", index=False)
account_summary.to_csv("account_summary.csv")

print("\nANALYSIS COMPLETED SUCCESSFULLY!")
print("Summary CSV files have been created.")
```
